```python
def preprocess_telco_data(df):
    # Tworzymy kopię, żeby nie psuć oryginału
    df_processed = df.copy()
    
    # ---------------------------------------------------------
    # 1. ZMIENNE BINARNE (0 i 1)
    # ---------------------------------------------------------
    # Lista kolumn, które mają tylko 'Yes'/'No'
    binary_cols = ['Partner', 'Dependents', 'PhoneService', 
                   'PaperlessBilling', 'Churn']
    
    # Zamieniamy pętlą, bo logika jest ta sama
    for col in binary_cols:
        df_processed[col] = df_processed[col].map({'Yes': 1, 'No': 0})
        
    # Gender traktujemy oddzielnie (bo nie ma Yes/No)
    df_processed['gender'] = df_processed['gender'].map({'Female': 1, 'Male': 0})

    # ---------------------------------------------------------
    # 2. ZMIENNE PORZĄDKOWE (Ordinal)
    # ---------------------------------------------------------
    # Tu definiujemy hierarchię
    contract_mapping = {
        'Month-to-month': 0, 
        'One year': 1, 
        'Two year': 2
    }
    df_processed['Contract'] = df_processed['Contract'].map(contract_mapping)

    # ---------------------------------------------------------
    # 3. ZMIENNE NOMINALNE (One-Hot Encoding)
    # ---------------------------------------------------------
    # Lista kolumn, które nie mają kolejności, a mają więcej niż 2 opcje
    nominal_cols = ['PaymentMethod', 'InternetService', 'DeviceProtection', 
                    'TechSupport', 'StreamingTV', 'StreamingMovies', 
                    'OnlineBackup', 'OnlineSecurity', 'MultipleLines']
    
    # get_dummies robi nowe kolumny i usuwa stare. 
    # drop_first=True jest kluczowe (wyjaśnienie pod kodem)
    df_processed = pd.get_dummies(df_processed, 
                                  columns=nominal_cols, 
                                  drop_first=True, 
                                  dtype=int) # dtype=int wymusza 0/1 zamiast True/False
    
    return df_processed

```

```python
df_clean = preprocess_telco_data(df2)
df_clean.head()
```
